## ⚙️ What is Data Preprocessing? ##

Data Preprocessing is the process of preparing raw data for analysis or machine learning. It involves cleaning and transforming data to improve its quality and usability 

### Common tasks ###
- Handling missing values
- Removing duplicates
- Converting data types
- Feature Scaling
- Encoding categorical data
- Function Transformer

____

## 🔄 What is Data Wrangling? ##

Data Wrangling is the process of transforming and restructing data into a desired format for analysis. It typically happens during the exploration stage and focuses on reshaping and understanding the dataset.

### Common tasks ###
- Merging or joining datasets
- Filtering and sorting data
- Creating new features
- Reshaping data (e.g., pivoting)

____

## ✅ Benefits of Cleaned Data ##

- Improved accuracy of analysis and models
- Better decision-making based on reliable data
- Smoother and faster data processing
- Clearer visualizations and summaries
- Enhanced stakeholder trust
- Reduced storage and maintenance issues

___

## ⚠️ Limitations and Considerations ##

Even with cleaned and well-prepared data, certain challenges remain:

- Misinterpretation of patterns due to coincidental or spurious correlations
- Hidden biases in data collection or labeling
- Overfitting to training data, especially when noise is present
- Visualizations or aggregations that obscure true trends
- Contextual errors—data may be technically clean but misleading without proper domain context

> Additionally, models may perform exceptionally well during training due to **unintended data leakage** — when information from the target variable or future events accidentally influences model input. This results in overly optimistic metrics that won't hold in real-world scenarios.

🏗️ Always validate assumptions, use proper data splits, and avoid including target-related features during training to prevent these pitfalls.
