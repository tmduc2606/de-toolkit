#### There are two primary ways to remove outliers: IQR and Z-score ####

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("loan_data_set.csv")
df.head(5)
```

```python
df.info()
```

```python
#Visualisation of outliers by Box Plot
sns.boxplot(x = "ApplicantIncome", data = df)
```

```python
#Alternative to distplot (deprecated in v0.14)
sns.displot(x = "ApplicantIncome", data = df, kde = True)
```
