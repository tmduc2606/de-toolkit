<b>Categorical data</b> refers to variables that represent <b>distinct groups or categories</b>. Each of value section is classified under those variables that are called discrete labels. These labels do not have numerical meaning, <b>qualities over quantities</b>.

Types of categorical data: <b>Nominal</b> (no levels or orders) - Male, PaymentMethod, Genders, Color; <b>Ordinal</b> (ordered sections) - Priority (High, Medium, Low), Ratings: Good, Bad, Mediocre...

```python
df = pd.read_csv("marvel-wikia-data.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
missing_cols = ["GSM", "APPEARANCES", "FIRST APPEARANCE"]

df[df[missing_cols].isnull().all(axis = 1)]
```

```python
#Filling NaN values for specific column
#df["SEX"].fillna(df["SEX"].mode()[0])
#df
```

```python
#Filling NaN values for mixed-types
str_cols = df.select_dtypes(include = "string").columns
num_cols = df.select_dtypes(include = ["float", "int64", "int32"]).columns

df[str_cols] = df[str_cols].fillna("None")
df[num_cols] = df[num_cols].fillna(0) 

df
```

```python
df.info()
```

```python
#Forward filling and backward filling
df.ffill()
```

```python
df.bfill()
```

```python
#Validating NaN values with selective datatypes

df.select_dtypes(include = "object").isnull()
```
