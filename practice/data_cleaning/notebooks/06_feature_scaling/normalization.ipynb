### Normalization ###

A scaling technique in which values are shifted and rescaled so that they end up ranging between 0 and 1, while this is much similar to Min - Max scaling, the difference lies in which horizontal types they are using (Normalization : x_mean & Min-max : x_min)

Normalization is particularly useful when the distribution of the data is unknown or does not follow a Gaussian distribution.
