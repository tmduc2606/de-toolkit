## What is encoding? ##

Encoding is a process of converting categorical data groups into numerical representations that datasets are usable by Machine Learning algorithms. Depending on the essence of the variable (e.g: ordinal or nominal...) and qualities such as cardinality, different encoding strategies are used. Each methods has strengths and weaknesses based on the data-context.

## Groups Of Encoding Strategy ##

<li> Unordered / Norminal (without ordering): One-hot, Frequency/Count, Binary, Hashing
<li> Ordered / Ordinal (with order): Label / Ordinal
<li> Supervised (target variable): Target / Mean, Leave-One-Out
