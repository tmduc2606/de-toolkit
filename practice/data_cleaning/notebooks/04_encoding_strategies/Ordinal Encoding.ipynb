```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv("loan_data_set.csv")
df.head(10)
```

```python
#Handling missing NaN values in ordinal columns
df = df.assign(Property_Area=lambda d: d["Property_Area"].fillna(d["Property_Area"].mode()[0]))
```

```python
ordinal_urban = [["Rural", "Semiurban", "Urban"]]
```

```python
oen = OrdinalEncoder(categories = ordinal_urban)
df["Property_Area"] = oen.fit_transform(df[["Property_Area"]])
```
