## 🧹 Introduction to Data Cleaning 📊 ##

### Definition ###

Data cleaning is a process of preparing data for analysis, further training in model Machine Learning / Deep Learning. This phase is being done by removing, modifying data that is incorrect, incomplete, irrelevant, duplicated or improperly formatted.

### Core skills ###

<li> Data Exploration 
<li> Handling Anomalies: Missing Values and Outliers 
<li> Filtering and Subsetting Data 
<li> RegEx and String Manipulation 
<li> Date/Time Parsing 
<li> Merging and Joining Datasets 
<li> Data Transformation and Data Type Conversions 
<li> Data Integrity: Deduplication and Validation 
<li> Dealing with Sparse Data 
<li> Normalization and Standardization 

### Additional topics ###

<li> Data Audit Reports (automated profiling with pandas-profiling, sweetviz)
<li> Logging Data Cleaning Decisions (to ensure reproducibility)
<li> Feature Engineering Preview (creating new useful features from existing ones)
