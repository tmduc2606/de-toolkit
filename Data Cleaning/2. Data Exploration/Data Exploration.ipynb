## First Impression ##

This dataset compiles all Marvel figures (played by actors) in the universe, provided with URL slug name for convenient Wikipedia search. 

Viewers presumably have impression on good characters, who are determined to save the world, bear prolonged imprints toward children, whereas Neutral and bad characters are often considered to be "evil".

However, there exists an imbalance in terms of genders, where male figures have tendency to shine other than female ones such as Carol Danvers.

Some of basic identities of a Marvel figure: Name, ID, Align (Heroes - Neutral - Villians), Eye, Hair, Sex, GSM, Alive, Appearances, Year - First Appearances
