```python
#Convert from categorical data into numerical groups (In this case we have Genders and Married are two categorical ones)
categorical_columns = ["Gender", "Married", "Education", "Self_Employed", "Loan_Status"]
encoded_data = df[categorical_columns]
encoded_data.fillna("None")
```

```python
#Alternative to Pandas: Sci-kit Learn, One-hot encoders
from sklearn.preprocessing import OneHotEncoder
```

By adding drop = "first" into One Hot Encoder function, we should avoid the Dummy Variable Trap (Redundancy and Multicollinearity).

### Dummy Variable Trap ###
A situation where one feature can be perfectly predicted using the others. This causes problems in linear models (like Linear or Logistic Regression), where this kind of perfect multicollinearity can lead to:
- Incorrect coefficient estimation
- Model instability

#### "Unothordox - More is not always better - It's about the independece of the features, not just their numbers" ####

```python
#ohe = OneHotEncoder(drop = "first", sparse_output = False)
ohe = OneHotEncoder(sparse_output = False)
one_hot_encoded = ohe.fit_transform(encoded_data)
ohe_df = pd.DataFrame(one_hot_encoded, columns = ohe.get_feature_names_out(categorical_columns))
```

```python
df_sklearn_encoded = pd.concat([df.drop(categorical_columns, axis = 1), ohe_df], axis = 1)
df_sklearn_encoded
```
