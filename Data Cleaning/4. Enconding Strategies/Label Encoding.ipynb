```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("Iris.csv")
```

```python
#Summary of species
df["Species"].value_counts()
```

```python
label_column = df["Species"]
label_column
```

```python
#Using Label Encoder from Scikit-learn to transform
le = LabelEncoder()
encoded_label = le.fit_transform(label_column)
le_df = pd.DataFrame(encoded_label, columns = ["Species"])
```

```python
df_updated = pd.concat([df.drop(columns = [label_column.name], axis = 1), le_df], axis = 1)
df_updated
```
