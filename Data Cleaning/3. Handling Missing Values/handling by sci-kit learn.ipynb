```python
from sklearn.impute import SimpleImputer
```

### There are four primary strategies to fill in NaN values: mean, median, most_frequent, constant 


| Strategy        | Applies To                | Description                                                   | Common Use Case                                         |
|----------------|---------------------------|---------------------------------------------------------------|---------------------------------------------------------|
| `mean`         | Numerical                 | Fills missing values with the **mean** of the column           | When numerical data is roughly **normally distributed** |
| `median`       | Numerical (robust to outliers) | Fills missing values with the **median** of the column    | When data contains **outliers**                         |
| `most_frequent`| Numerical & Categorical   | Fills missing values with the **most frequent value** (mode)   | For **categorical** features or simple numerics         |
| `constant`     | Numerical & Categorical   | Fills missing values with a **constant** (set via `fill_value`) | To use placeholders like `'Unknown'`, `0`, or `-999`    |


```python
si = SimpleImputer(strategy = "mean")
ar = si.fit_transform((df[['RATING', 'RunTime']]))
```

```python
#Displaying the filtered categories by displaying as DataFrame with the index column selected with dtypes - float64
new_df = pd.DataFrame(ar, columns = df.select_dtypes(include = "float64").columns)
new_df
```
