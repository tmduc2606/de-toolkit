```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("marvel-wikia-data.csv")
```

```python
#Check which figures contain most NaN identities

missing_cols = ["APPEARANCES", "Year", "FIRST APPEARANCE"]
df[df[missing_cols].isna().all(axis = 1)]
```

```python
#Analyzing which columns get most NaN values (Percentage)

(df.isnull().sum() / df.shape[0]) * 100
```

```python
#Total of NaN columns
df.isnull().sum().sum()
```

```python
#Total NaN values of dataset
(df.isnull().sum().sum()) / (df.shape[0] * df.shape[1]) * 100

#This infers that the dataset is significantly challenging to manage, especially the process of cleaning will become much harsher
```

```python
#Total of non-NaN values
df.notnull().sum()
```

```python
#Visualisation of missing values w/ Heatmap

sns.heatmap(df.isnull())
plt.show()
```
